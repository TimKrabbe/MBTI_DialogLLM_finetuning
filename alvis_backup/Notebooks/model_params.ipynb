# Model Params

During fine-tuning I forgot to print the number of trainable params of my LoRA adapter, so I do it here.

```python
from peft import PeftModel
from transformers import AutoModelForCausalLM
```

```python
base_model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen2.5-7B-Instruct")
model = PeftModel.from_pretrained(base_model, "DrinkIcedT/Qwen2.5-7B-Instruct_MBTI-Dialogues_modifiedprompt")

# adapter params filtered by name
adapter_params = sum(
    p.numel() for name, p in model.named_parameters()
    if "lora_" in name
)
total_params = sum(p.numel() for p in model.parameters())

print(f"LoRA params:  {adapter_params:,}")
print(f"Total params: {total_params:,}")
print(f"Trainable %:  {100 * adapter_params / total_params:.4f}%")
```
